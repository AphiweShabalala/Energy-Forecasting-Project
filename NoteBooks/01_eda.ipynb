```python
col = 'MT_178'

zero_mask = df[col] == 0

print("Total observations:", len(df))
print("Zero observations:", zero_mask.sum())
print("Non-zero observations:", (~zero_mask).sum())
```

```python
non_zero_times = df.index[df[col] != 0]

print("First non-zero:", non_zero_times.min())
print("Last non-zero:", non_zero_times.max())
```

```python
hourly_zero_rate = (
    df[col]
    .groupby(df.index.hour)
    .apply(lambda x: (x == 0).mean() * 100)
)

hourly_zero_rate
```

```python
hourly_zero_rate.plot(
    kind='bar',
    figsize=(12, 5)
)

plt.xlabel('Hour of Day')
plt.ylabel('Zero Percentage')
plt.title(f'Zero Rate by Hour — {col}')
plt.xticks(rotation=0)
plt.show()
```

```python
weekday_zero_rate = (
    df[col]
    .groupby(df.index.dayofweek)
    .apply(lambda x: (x == 0).mean() * 100)
)

weekday_zero_rate
```

```python
weekday_zero_rate.plot(
    kind='bar',
    figsize=(10, 5)
)

plt.xlabel('Day of Week')
plt.ylabel('Zero Percentage')
plt.title(f'Zero Rate by Day of Week — {col}')
plt.xticks(
    range(7),
    ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'],
    rotation=0
)
plt.show()
```

```python
monthly_zero_rate = (
    df[col]
    .groupby(df.index.month)
    .apply(lambda x: (x == 0).mean() * 100)
)

monthly_zero_rate
```

```python
monthly_zero_rate.plot(
    kind='bar',
    figsize=(10, 5)
)

plt.xlabel('Month')
plt.ylabel('Zero Percentage')
plt.title(f'Zero Rate by Month — {col}')
plt.xticks(range(12), rotation=0)
plt.show()
```

```python
hourly = df.resample('1h').mean()
hourly[['MT_001', 'MT_370']].loc['2014-10-01':'2014-10-08'].plot(
    figsize=(16, 6)
)

plt.xlabel('Time')
plt.ylabel('Average load (kW)')
plt.title('Hourly Electricity Load')
plt.tight_layout()
plt.show()
```

```python
# Average load across all clients at each timestamp
total_load = df.sum(axis=1)

# Average load for each hour of the day
hourly_pattern = total_load.groupby(total_load.index.hour).mean()

# Average load for each day of the week
weekly_pattern = total_load.groupby(total_load.index.dayofweek).mean()

# Plot both patterns
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

hourly_pattern.plot(ax=axes[0], marker='o')
axes[0].set_title('Average Electricity Load by Hour')
axes[0].set_xlabel('Hour of Day')
axes[0].set_ylabel('Average Total Load (kW)')
axes[0].set_xticks(range(0, 24, 2))

weekly_pattern.index = [
    'Monday', 'Tuesday', 'Wednesday', 'Thursday',
    'Friday', 'Saturday', 'Sunday'
]

weekly_pattern.plot(kind='bar', ax=axes[1])
axes[1].set_title('Average Electricity Load by Day of Week')
axes[1].set_xlabel('Day')
axes[1].set_ylabel('Average Total Load (kW)')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```
